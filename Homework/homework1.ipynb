Q6. Median value of horsepower

```python
# 1. Find the median value of the horsepower column in the dataset

median_horsepower = df.horsepower.median()

print(f"The median value of the horsepower column in the dataset is {median_horsepower}")
```

```python
# 2. Calculate the most frequent value of the same horsepower column

mode_horsepower = df.horsepower.mode()[0]

print(f"The most frequent value of the horsepower column in the dataset is {mode_horsepower}")
```

```python
# 3. Use the fillna method to fill the missing values in the horsepower column with the most frequent value from the previous step

df.horsepower = df.horsepower.fillna(mode_horsepower)
```

```python
# 4. Calculate the median value of horsepower once again.

median_horsepower_filled = df.horsepower.median()

print(f"The median value of horsepower after filling the missing values with the most frequent value is {median_horsepower_filled}")
```

***The median value neither increased nor decreased after filling the missing values in the horsepower column with the most frequent value.***

# Q7. Sum of weights

```python
# Select all the cars from Asia only vehicle_weight and model_year columns and the first 7 values

df_asia = df[df.origin=='Asia'][['vehicle_weight', 'model_year']].head(7)
df_asia
```

```python
# Get the underlying NumPy array. Let's call it X

X = df_asia.values
X
```

```python
# Compute matrix-matrix multiplication between the transpose of X and X. To get the transpose, use X.T. Let's call the result XTX
# Use functions for this task: complete the task. Don't stop at the function definition.Remember that Y should be the transpose of X, which is XT

XT = X.T
XT

XTX = np.dot(XT, X)
XTX
```

```python
# Invert XTX

XTX_inv = np.linalg.inv(XTX)
XTX_inv
```

```python
# Create an array y with values [1100, 1300, 800, 900, 1000, 1100, 1200]

y = np.array([1100, 1300, 800, 900, 1000, 1100, 1200])
y
```

```python
# Multiply the inverse of XTX with the transpose of X, and then multiply the result by y. Call the result w.

w = np.dot(np.dot(XTX_inv, XT), y)
print('w =', w)
```
